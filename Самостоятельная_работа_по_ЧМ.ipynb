Задание 1

```python
import numpy as np
import matplotlib.pyplot as plt

def f1(x):
  return 2*np.cos(x)
def f2(x):
  return 2**x

x1 = np.linspace(-10,5, 10000)
y1 = f1(x1)
x2 = np.linspace(-10,5,1000)
y2 = f2(x2)

plt.plot(x1, y1)
plt.plot(x2, y2)
plt.axhline(y=0, color='k', linestyle='-')
plt.axvline(x=0, color='k', linestyle='-')
plt.grid(True)
plt.title('График f(x)=2cos(x) и f(x)=2^x')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.show()
```

```python
from math import *

def f(x):
  return 2**x - 2*cos(x)

epselant = 1
a = 0
b = 1
while epselant > 0.001:
  c = (a + b)/2
  if f(a) * f(c) < 0:
    b = c
  else:
    a = c
  epselant = (b - a)/2
globalAnswer = (a + b)/2
print(f'Ответ: {round(globalAnswer,3)}')
```

Задание 3

```python
from math import *

def f(x):
  return acos(2**x/2)

x_n1 = -1
x_n2 = -1.4
count = 1
epselant = 10**(-6)
print(f'0 --- {x_n2}')
while abs(x_n2 - x_n1) > epselant:
    x_n1 = x_n2
    x_n2 = round(f(x_n1), 6)
    print(f'{count} --- {x_n2}')
    count += 1
print(f'ANSWER {x_n2}')
```
